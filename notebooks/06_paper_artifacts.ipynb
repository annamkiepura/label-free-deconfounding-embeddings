# 06 · The paper's tables and figures

Builds every table and figure of the paper from the saved results (`results/`) with `paper-artifacts/build.py`;
nothing is fitted. Writes `paper-artifacts/tables/*.tex` (and Markdown copies), `paper-artifacts/figures/*.pdf`,
the data behind each exhibit with its provenance (`paper-artifacts/data/`) and a build report
(`paper-artifacts/reports/build_report.md`). Every exhibit records the checks it ran; the build fails on any error.
A few minutes on a laptop (the first build also computes the frozen-transfer figure's five-seed bootstrap summaries).
